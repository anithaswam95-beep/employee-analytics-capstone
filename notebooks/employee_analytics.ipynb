````sql
-- Clean temp views to reset data for this session (shadows the tables without modifying them)

CREATE OR REPLACE TEMP VIEW employees AS
SELECT * FROM VALUES
  (1, 'John',   'IT',      80000),
  (2, 'Mary',   'HR',      70000),
  (3, 'David',  'IT',      90000),
  (4, 'Lisa',   'Finance', 85000),
  (5, 'Robert', 'Sales',   65000),
  (6, 'Sarah',  'HR',      75000),
  (7, 'Tom',    NULL,      60000)
AS t(employee_id, name, department, salary);

CREATE OR REPLACE TEMP VIEW departments AS
SELECT * FROM VALUES
  ('IT', 'David'),
  ('HR', 'Lisa'),
  ('Sales', 'Robert'),
  ('Finance', 'Lisa')
AS t(department, manager);

SELECT 'employees' AS table_name, COUNT(*) AS row_count FROM employees
UNION ALL
SELECT 'departments', COUNT(*) FROM departments;
````

````sql
    
-- GROUP BY: Group rows by one or more columns
-- Aggregate functions: COUNT, SUM, AVG, MIN, MAX

SELECT 
    department,
    COUNT(*) AS employee_count,
    AVG(salary) AS avg_salary,
    MAX(salary) AS max_salary,
    MIN(salary) AS min_salary
FROM employees
GROUP BY department;
````

````sql
%sql
-- ORDER BY: Sort results by one or more columns
-- LIMIT: Restrict the number of rows returned

SELECT employee_id, name, department, salary
FROM employees
ORDER BY salary DESC
LIMIT 3;
````

````sql
    
-- HAVING: Filter on aggregated values (like WHERE but for GROUP BY)

SELECT 
    department,
    COUNT(*) AS employee_count,
    AVG(salary) AS avg_salary
FROM employees
GROUP BY department
HAVING COUNT(*) > 1
ORDER BY avg_salary DESC;
````

````sql
    
-- LEFT JOIN: Returns all rows from the left table,
-- and matched rows from the right table. Unmatched rows get NULL.

SELECT e.employee_id, e.name, e.department, e.salary, d.manager
FROM employees e
LEFT JOIN departments d
ON e.department = d.department
ORDER BY e.employee_id;
````

````sql
    
-- RIGHT JOIN: Returns all rows from the right table (departments),
-- and matched rows from the left table (employees).
-- Departments with no employees still appear, with NULL for employee columns.

SELECT e.employee_id, e.name, e.department, e.salary, d.manager
FROM employees e
RIGHT JOIN departments d
ON e.department = d.department
ORDER BY d.department, e.employee_id;
````

````sql
    
-- FULL OUTER JOIN: Returns all rows from BOTH tables.
-- Matched rows are joined; unmatched rows from either side appear with NULLs.
-- Useful for finding gaps in relationships (e.g., employees without a department,
-- or departments with no employees).

SELECT e.employee_id, e.name, e.department, e.salary, d.manager
FROM employees e
FULL OUTER JOIN departments d
ON e.department = d.department
ORDER BY e.employee_id, d.department;
````

````sql
-- DISTINCT: Remove duplicate rows from the result set

SELECT DISTINCT department
FROM employees
ORDER BY department;
````

````sql
-- WHERE with operators: IN, BETWEEN, LIKE

-- IN: Match any value in a list
SELECT employee_id, name, department, salary
FROM employees
WHERE department IN ('IT', 'Finance')
ORDER BY salary DESC;

-- BETWEEN: Filter by a range (inclusive)
SELECT employee_id, name, salary
FROM employees
WHERE salary BETWEEN 70000 AND 85000
ORDER BY salary;

-- LIKE: Pattern matching with wildcards
-- % = any sequence of characters, _ = single character
SELECT employee_id, name
FROM employees
WHERE name LIKE '%a%';
````

````sql
-- IS NULL / IS NOT NULL: Handle missing values
-- Tom (employee 7) already has a NULL department in our dataset

-- Find employees with no department assigned
SELECT employee_id, name, department, salary
FROM employees
WHERE department IS NULL;

-- Find employees who DO have a department
SELECT employee_id, name, department, salary
FROM employees
WHERE department IS NOT NULL
ORDER BY employee_id;
````

````sql
-- COALESCE: Return the first non-NULL value from a list
-- Useful for providing defaults when data is missing

SELECT 
    employee_id,
    name,
    COALESCE(department, 'Unassigned') AS department,
    salary
FROM employees
ORDER BY employee_id;
````

````sql
-- CASE WHEN: Conditional logic in SQL (like if/else)
-- Classify employees by salary tier

SELECT 
    employee_id,
    name,
    department,
    salary,
    CASE 
        WHEN salary >= 85000 THEN 'High'
        WHEN salary >= 70000 THEN 'Medium'
        ELSE 'Low'
    END AS salary_tier
FROM employees
ORDER BY salary DESC;
````

````sql
-- UNION / UNION ALL / INTERSECT / EXCEPT: Combine result sets

-- UNION: Combine rows and remove duplicates
SELECT name FROM employees WHERE department = 'IT'
UNION
SELECT name FROM employees WHERE department = 'HR'
ORDER BY name;

-- UNION ALL: Combine rows keeping duplicates (faster, no dedup)
SELECT 'IT' AS src, name FROM employees WHERE department = 'IT'
UNION ALL
SELECT 'HR' AS src, name FROM employees WHERE department = 'HR'
ORDER BY src, name;

-- INTERSECT: Rows present in BOTH result sets
SELECT department FROM employees
INTERSECT
SELECT department FROM departments;

-- EXCEPT: Rows in the first set but NOT in the second
SELECT department FROM departments
EXCEPT
SELECT department FROM employees;
````

````sql
-- Subqueries: Query inside a query

-- Scalar subquery: returns a single value
SELECT employee_id, name, salary
FROM employees
WHERE salary > (SELECT AVG(salary) FROM employees)
ORDER BY salary DESC;

-- IN subquery: filter using a list returned by a subquery
SELECT employee_id, name, department
FROM employees
WHERE department IN (SELECT department FROM departments WHERE manager = 'Lisa')
ORDER BY name;

-- EXISTS subquery: check if related rows exist
SELECT d.department, d.manager
FROM departments d
WHERE EXISTS (
    SELECT 1 FROM employees e
    WHERE e.department = d.department
)
ORDER BY d.department;

-- Correlated subquery: subquery references the outer row
SELECT e.employee_id, e.name, e.salary,
       e.salary - (SELECT AVG(salary) FROM employees e2
                    WHERE e2.department = e.department) AS diff_from_dept_avg
FROM employees e
WHERE e.department IS NOT NULL
ORDER BY e.department, e.employee_id;
````

````sql
-- Self-Join: Join a table to itself
-- Find employees who share the same department

SELECT a.employee_id AS emp1_id, a.name AS emp1_name,
       b.employee_id AS emp2_id, b.name AS emp2_name,
       a.department
FROM employees a
JOIN employees b
  ON a.department = b.department
  AND a.employee_id < b.employee_id
ORDER BY a.department, a.employee_id;

-- CROSS JOIN: Every row from the left table paired with every row from the right
-- (Cartesian product)
SELECT e.name AS employee, d.department AS dept_option
FROM employees e
CROSS JOIN departments d
WHERE e.department IS NOT NULL
ORDER BY e.employee_id, d.department;
````

````sql
-- CTE (Common Table Expression): Named subquery defined with WITH
-- More readable than nested subqueries and can be referenced multiple times

-- Simple CTE: high earners
WITH high_earners AS (
    SELECT employee_id, name, department, salary
    FROM employees
    WHERE salary >= 75000
)
SELECT department, COUNT(*) AS count, AVG(salary) AS avg_salary
FROM high_earners
GROUP BY department
ORDER BY avg_salary DESC;

-- Multiple CTEs chained together
WITH dept_stats AS (
    SELECT department, AVG(salary) AS dept_avg
    FROM employees
    WHERE department IS NOT NULL
    GROUP BY department
),
above_avg AS (
    SELECT e.employee_id, e.name, e.department, e.salary
    FROM employees e
    JOIN dept_stats d ON e.department = d.department
    WHERE e.salary > d.dept_avg
)
SELECT * FROM above_avg
ORDER BY salary DESC;
````

## Advanced SQL

Window functions, string & date manipulation, and pivoting data.

````sql
-- Window Functions: Perform calculations across a set of rows related to the current row
-- Unlike GROUP BY, window functions keep all rows in the output

-- ROW_NUMBER: Assign a unique sequential number to each row within a partition
SELECT
    employee_id, name, department, salary,
    ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS row_num
FROM employees
WHERE department IS NOT NULL
ORDER BY department, salary DESC;

-- RANK vs DENSE_RANK: Ranking with ties
-- RANK: skips numbers after ties (1, 1, 3)
-- DENSE_RANK: no gaps after ties (1, 1, 2)
SELECT
    employee_id, name, department, salary,
    RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS rank,
    DENSE_RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS dense_rank
FROM employees
WHERE department IS NOT NULL
ORDER BY department, salary DESC;

-- Running total with SUM OVER: cumulative salary per department
SELECT
    employee_id, name, department, salary,
    SUM(salary) OVER (PARTITION BY department ORDER BY salary DESC) AS running_total
FROM employees
WHERE department IS NOT NULL
ORDER BY department, salary DESC;

-- LAG / LEAD: Access the previous / next row's value
SELECT
    employee_id, name, department, salary,
    LAG(salary, 1) OVER (PARTITION BY department ORDER BY salary DESC) AS prev_salary,
    LEAD(salary, 1) OVER (PARTITION BY department ORDER BY salary DESC) AS next_salary
FROM employees
WHERE department IS NOT NULL
ORDER BY department, salary DESC;
````

````sql
-- String Functions: Manipulate text data

SELECT
    name,
    UPPER(name)         AS upper_name,
    LOWER(name)         AS lower_name,
    LENGTH(name)        AS name_length,
    SUBSTRING(name, 1, 3) AS first_3_chars,
    CONCAT(name, ' - ', COALESCE(department, 'Unassigned')) AS name_with_dept,
    TRIM('  hello  ')   AS trimmed
FROM employees
ORDER BY employee_id
LIMIT 5;
````

````sql
-- Date Functions: Work with dates and times

-- Current date and time
SELECT
    CURRENT_DATE()        AS today,
    CURRENT_TIMESTAMP()   AS now;

-- Date arithmetic: add/subtract days, compute differences
SELECT
    CURRENT_DATE()                       AS today,
    DATE_ADD(CURRENT_DATE(), 30)         AS plus_30_days,
    DATE_SUB(CURRENT_DATE(), 7)          AS minus_7_days,
    DATEDIFF(CURRENT_DATE(), DATE '2026-01-01') AS days_since_jan1,
    DATE_FORMAT(CURRENT_DATE(), 'yyyy-MM-dd') AS formatted_date;

-- Date parts: extract components
SELECT
    CURRENT_DATE()              AS full_date,
    YEAR(CURRENT_DATE())        AS yr,
    MONTH(CURRENT_DATE())       AS mon,
    DAY(CURRENT_DATE())         AS day,
    DAYNAME(CURRENT_DATE())     AS day_name;
````

````sql
-- PIVOT / UNPIVOT: Reshape data between rows and columns

-- PIVOT: Rotate rows into columns
-- Show employee count per department, with departments as columns

-- First, create a temp view for pivoting
CREATE OR REPLACE TEMP VIEW emp_pivot_source AS
SELECT department, salary, employee_id
FROM employees
WHERE department IS NOT NULL;

SELECT * FROM emp_pivot_source
PIVOT (
    COUNT(employee_id) AS emp_count
    FOR department IN ('IT' AS it_count, 'HR' AS hr_count, 'Finance' AS fin_count, 'Sales' AS sales_count)
);

-- UNPIVOT: Rotate columns back into rows
-- Using a small sample table to demonstrate
CREATE OR REPLACE TEMP VIEW sales_by_region AS
SELECT * FROM VALUES
    ('Q1', 100, 200, 150),
    ('Q2', 120, 180, 160),
    ('Q3', 130, 210, 170)
AS t(quarter, north, south, west);

SELECT * FROM (
    SELECT * FROM sales_by_region
)
UNPIVOT (
    sales_amount FOR region IN (north, south, west)
)
ORDER BY quarter, region;
````

## More Advanced SQL

Data modification (DML), multi-dimensional aggregations, QUALIFY, regular expressions, type conversions, arrays, and JSON/VARIANT handling.

````sql
-- UPDATE and DELETE: Modify existing rows in a table
-- Temp views don't support DML in Unity Catalog, so we create a demo table via CTAS

CREATE TABLE IF NOT EXISTS sql_demo_update AS SELECT * FROM employees;

-- UPDATE: Change values in existing rows
UPDATE sql_demo_update
SET salary = salary + 5000
WHERE department = 'IT';

SELECT * FROM sql_demo_update WHERE department = 'IT' ORDER BY employee_id;

-- DELETE: Remove rows matching a condition
DELETE FROM sql_demo_update
WHERE department IS NULL;

SELECT 'After DELETE:' AS note, COUNT(*) AS row_count FROM sql_demo_update;
````

````sql
-- MERGE INTO: Upsert — insert new rows, update existing ones, or delete matched rows
-- Combines INSERT, UPDATE, and DELETE in a single statement

CREATE TABLE IF NOT EXISTS sql_demo_merge AS SELECT * FROM employees;

CREATE OR REPLACE TEMP VIEW employee_updates AS
SELECT * FROM VALUES
  (1, 'John',   'IT',       82000),   -- existing → will be UPDATE'd
  (3, 'David',  'IT',       92000),   -- existing → will be UPDATE'd
  (8, 'Anna',   'Marketing', 78000)   -- new → will be INSERT'ed
AS t(employee_id, name, department, salary);

MERGE INTO sql_demo_merge AS t
USING employee_updates AS s
  ON t.employee_id = s.employee_id
WHEN MATCHED THEN
  UPDATE SET salary = s.salary, department = s.department
WHEN NOT MATCHED THEN
  INSERT (employee_id, name, department, salary) VALUES (s.employee_id, s.name, s.department, s.salary);

SELECT * FROM sql_demo_merge ORDER BY employee_id;
````

````sql
-- GROUPING SETS / ROLLUP / CUBE: Multi-dimensional aggregations
-- GROUP BY gives one level of aggregation; these give multiple levels at once

-- ROLLUP: Hierarchical subtotals (department → grand total)
SELECT
    COALESCE(department, 'ALL') AS department,
    COUNT(*) AS emp_count,
    SUM(salary) AS total_salary
FROM employees
WHERE department IS NOT NULL
GROUP BY ROLLUP (department)
ORDER BY department;

-- CUBE: All possible combinations of grouping columns
-- With one column, CUBE = ROLLUP; with two+ columns it generates every subset
SELECT
    COALESCE(department, 'ALL') AS department,
    CASE WHEN GROUPING(department) = 1 THEN 'Subtotal' ELSE 'Detail' END AS row_type,
    COUNT(*) AS emp_count,
    AVG(salary) AS avg_salary
FROM employees
WHERE department IS NOT NULL
GROUP BY CUBE (department)
ORDER BY row_type, department;

-- GROUPING SETS: Specify exact grouping levels you want
SELECT
    COALESCE(department, 'ALL') AS department,
    COUNT(*) AS emp_count
FROM employees
WHERE department IS NOT NULL
GROUP BY GROUPING SETS ((department), ())
ORDER BY department;
````

````sql
-- QUALIFY: Filter on the results of window functions
-- Like HAVING but for window functions — avoids wrapping the query in a subquery

-- Find the highest-paid employee in each department
SELECT
    employee_id, name, department, salary,
    RANK() OVER (PARTITION BY department ORDER BY salary DESC) AS salary_rank
FROM employees
WHERE department IS NOT NULL
QUALIFY RANK() OVER (PARTITION BY department ORDER BY salary DESC) = 1
ORDER BY salary DESC;

-- Top 2 earners per department
SELECT
    employee_id, name, department, salary,
    ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) AS rn
FROM employees
WHERE department IS NOT NULL
QUALIFY ROW_NUMBER() OVER (PARTITION BY department ORDER BY salary DESC) <= 2
ORDER BY department, salary DESC;
````

````sql
-- Regular Expressions: Pattern matching with powerful regex syntax

-- REGEXP_LIKE: Check if a string matches a pattern
SELECT name, department,
       REGEXP_LIKE(name, '^[A-D]') AS starts_with_a_to_d
FROM employees
ORDER BY employee_id;

-- REGEXP_EXTRACT: Pull a matching substring
SELECT name,
       REGEXP_EXTRACT(name, '([aeiou])', 1) AS first_vowel
FROM employees
WHERE REGEXP_LIKE(name, '[aeiou]')
ORDER BY employee_id;

-- REGEXP_REPLACE: Replace matched patterns
SELECT name,
       REGEXP_REPLACE(name, '[aeiou]', '*') AS vowels_masked
FROM employees
ORDER BY employee_id;
````

````sql
-- Type Conversions: CAST and TRY_CAST
-- CAST fails on bad data; TRY_CAST returns NULL instead of erroring

SELECT
    name,
    salary,
    CAST(salary AS DOUBLE)        AS salary_double,
    CAST(salary AS STRING)        AS salary_string,
    TRY_CAST('abc' AS INT)        AS try_cast_bad_value,
    TRY_CAST('123' AS INT)        AS try_cast_good_value,
    CAST(CONCAT(salary, '.00') AS DECIMAL(10,2)) AS salary_decimal
FROM employees
ORDER BY employee_id
LIMIT 5;
````

````sql
-- ARRAY Functions: Work with array columns
-- Create a sample table with array data

CREATE OR REPLACE TEMP VIEW projects AS
SELECT * FROM VALUES
  (1, 'John',   ARRAY('Website', 'API', 'Migration')),
  (2, 'Mary',   ARRAY('HR Portal', 'Payroll')),
  (3, 'David',  ARRAY('Data Pipeline', 'API', 'ETL', 'Dashboard')),
  (4, 'Lisa',   ARRAY('Budget', 'Forecast'))
AS t(employee_id, name, projects);

-- ARRAY_SIZE: Count elements in an array
SELECT name, projects, ARRAY_SIZE(projects) AS num_projects
FROM projects
ORDER BY num_projects DESC;

-- EXPLODE: Turn each array element into its own row
SELECT name, EXPLODE(projects) AS project
FROM projects
ORDER BY name, project;

-- ARRAY_CONTAINS: Check if an array includes a value
SELECT name, ARRAY_CONTAINS(projects, 'API') AS works_on_api
FROM projects
ORDER BY employee_id;
````

````sql
-- JSON / VARIANT: Work with semi-structured data in Databricks
-- VARIANT is Databricks' unified type for JSON, Parquet, and other semi-structured data

CREATE OR REPLACE TEMP VIEW employee_json AS
SELECT * FROM VALUES
  (1, PARSE_JSON('{"name": "John", "skills": ["Python", "SQL"], "active": true}')),
  (2, PARSE_JSON('{"name": "Mary", "skills": ["Excel", "PowerBI"], "active": false}')),
  (3, PARSE_JSON('{"name": "David", "skills": ["Spark", "Scala", "Airflow"], "active": true}'))
AS t(id, profile);

-- Extract fields with : notation (VARIANT path access)
SELECT
    id,
    profile:name      AS name,
    profile:active    AS is_active,
    profile:skills    AS skills_array
FROM employee_json
ORDER BY id;

-- EXPLODE nested arrays inside JSON (cast VARIANT to ARRAY first)
SELECT
    id,
    profile:name AS name,
    EXPLODE(CAST(profile:skills AS ARRAY<STRING>)) AS skill
FROM employee_json
ORDER BY id, skill;

-- IS_JSON_STRING and TO_JSON for round-tripping
SELECT
    id,
    profile,
    TRY_CAST(profile AS STRING) AS json_string
FROM employee_json
ORDER BY id;
````

### Exercise 5: Find Peers (CTE + Self-Join)

**Requirements:**
- Return `emp1_name`, `emp2_name`, `department`, `emp1_salary`, `emp2_salary`, `salary_diff` (absolute difference)
- Use a CTE to filter to only employees with a department
- Self-join to find pairs where `emp1_id < emp2_id`
- Order by `salary_diff` descending

<details><summary>Hint</summary>Create a CTE `with_dept` that filters employees with non-NULL departments, then self-join it</details>

````sql
-- Solution: Exercise 5 — Find Peers (CTE + Self-Join)
WITH with_dept AS (
    SELECT employee_id, name, department, salary
    FROM employees
    WHERE department IS NOT NULL
)
SELECT 
    a.name AS emp1_name,
    b.name AS emp2_name,
    a.department,
    a.salary AS emp1_salary,
    b.salary AS emp2_salary,
    ABS(a.salary - b.salary) AS salary_diff
FROM with_dept a
JOIN with_dept b
  ON a.department = b.department
  AND a.employee_id < b.employee_id
ORDER BY salary_diff DESC;
````

### Exercise 6: Salary Gap Analysis (Subquery + String Functions + CONCAT)

- Only include employees with a department
- Order by department, then salary descending

````sql
-- Solution: Exercise 6 — Salary Gap Analysis
SELECT 
    e.name,
    e.department,
    e.salary,
    dept_max.max_salary - e.salary AS gap_from_top,
    CONCAT(e.name, ' earns ', dept_max.max_salary - e.salary, ' less than the top in ', e.department) AS summary
FROM employees e
JOIN (
    SELECT department, MAX(salary) AS max_salary
    FROM employees
    WHERE department IS NOT NULL
    GROUP BY department
) dept_max ON e.department = dept_max.department
WHERE e.department IS NOT NULL
ORDER BY e.department, e.salary DESC;
````

### Exercise 8: Comprehensive Challenge

Combine everything you've learned into a single query:

Write a query that produces a **department summary report** with the following columns:
- `department` — department name (use `COALESCE` for NULL → `'Unassigned'`)
- `emp_count` — number of employees
- `top_earner` — name of the highest-paid employee in that department
- `avg_salary` — average salary (rounded)
- `salary_tier` — `'Well-paid'` if avg ≥ 80K, `'Moderate'` if avg ≥ 70K, else `'Low-paid'`

**Requirements:**
- Use a CTE to compute per-department stats
- Use a window function (via QUALIFY or a subquery) to find the top earner
- Use CASE WHEN for the salary tier
- Order by avg_salary descending
- Include the `'Unassigned'` department (Tom's department is NULL)

````sql
-- Solution: Exercise 8 — Comprehensive Challenge
WITH dept_stats AS (
    SELECT
        COALESCE(department, 'Unassigned') AS department,
        COUNT(*) AS emp_count,
        ROUND(AVG(salary)) AS avg_salary
    FROM employees
    GROUP BY COALESCE(department, 'Unassigned')
),
top_earners AS (
    SELECT 
        COALESCE(department, 'Unassigned') AS department,
        name AS top_earner
    FROM employees
    QUALIFY ROW_NUMBER() OVER (
        PARTITION BY COALESCE(department, 'Unassigned') 
        ORDER BY salary DESC
    ) = 1
)
SELECT 
    d.department,
    d.emp_count,
    t.top_earner,
    d.avg_salary,
    CASE 
        WHEN d.avg_salary >= 80000 THEN 'Well-paid'
        WHEN d.avg_salary >= 70000 THEN 'Moderate'
        ELSE 'Low-paid'
    END AS salary_tier
FROM dept_stats d
JOIN top_earners t ON d.department = t.department
ORDER BY d.avg_salary DESC;
````

````sql
-- Recursive CTEs: Traverse hierarchical data (org charts, parts lists, graphs)
-- Uses WITH RECURSIVE with a base case + recursive step joined by UNION ALL

CREATE OR REPLACE TEMP VIEW org_chart AS
SELECT * FROM VALUES
  (1, 'CEO',              NULL),
  (2, 'VP Engineering',    1),
  (3, 'VP Sales',          1),
  (4, 'Eng Lead',          2),
  (5, 'Sales Rep',         3),
  (6, 'Developer',         4),
  (7, 'QA Engineer',       4)
AS t(employee_id, name, manager_id);

-- Traverse the org chart from CEO down to all reports
WITH RECURSIVE org_tree AS (
  -- Anchor: start with the CEO (no manager)
  SELECT employee_id, name, manager_id, 0 AS depth,
         CAST(name AS STRING) AS path
  FROM org_chart
  WHERE manager_id IS NULL

  UNION ALL

  -- Recursive: find direct reports of the previous level
  SELECT 
      e.employee_id,
      e.name,
      e.manager_id,
      t.depth + 1,
      CONCAT(t.path, ' > ', e.name)
  FROM org_chart e
  INNER JOIN org_tree t ON e.manager_id = t.employee_id
)
SELECT employee_id, name, depth, path
FROM org_tree
ORDER BY depth, employee_id;
````

````sql
-- Statistical Functions: Analyze distributions, correlations, and percentiles

-- Create sample data with two correlated variables
CREATE OR REPLACE TEMP VIEW sales_data AS
SELECT * FROM VALUES
  (1, 60, 200),
  (2, 65, 220),
  (3, 70, 250),
  (4, 75, 280),
  (5, 80, 310),
  (6, 85, 340),
  (7, 90, 380)
AS t(month, sales_volume, revenue);

-- CORR: Correlation between two columns (-1 to +1)
SELECT 
    CORR(sales_volume, revenue)     AS correlation,
    COVAR_SAMP(sales_volume, revenue) AS covariance,
    STDDEV(sales_volume)             AS stddev_volume,
    VARIANCE(revenue)               AS var_revenue
FROM sales_data;

-- PERCENTILE_CONT: Interpolated percentile (median, quartiles)
SELECT 
    PERCENTILE_CONT(0.5)  WITHIN GROUP (ORDER BY salary) AS median_salary,
    PERCENTILE_CONT(0.25) WITHIN GROUP (ORDER BY salary) AS p25_salary,
    PERCENTILE_CONT(0.75) WITHIN GROUP (ORDER BY salary) AS p75_salary
FROM employees;

-- APPROX_COUNT_DISTINCT: Fast approximate distinct count (HyperLogLog)
SELECT APPROX_COUNT_DISTINCT(department) AS approx_dept_count,
       COUNT(DISTINCT department)       AS exact_dept_count
FROM employees;
````

````sql
-- Conditional Aggregation: Aggregate different subsets in a single pass
-- Instead of multiple queries, use CASE WHEN inside aggregates or the FILTER clause

-- Approach 1: SUM(CASE WHEN ...) — pivot data into columns with one scan
SELECT 
    department,
    SUM(CASE WHEN salary >= 80000 THEN 1 ELSE 0 END) AS high_earners,
    SUM(CASE WHEN salary >= 70000 AND salary < 80000 THEN 1 ELSE 0 END) AS medium_earners,
    SUM(CASE WHEN salary < 70000 THEN 1 ELSE 0 END) AS low_earners,
    SUM(salary) AS total_salary
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY department;

-- Approach 2: FILTER clause — cleaner syntax for conditional aggregation
SELECT 
    department,
    COUNT(*)                                   AS total_employees,
    COUNT(*) FILTER (WHERE salary >= 80000)     AS high_earners,
    COUNT(*) FILTER (WHERE salary < 70000)      AS low_earners,
    AVG(salary) FILTER (WHERE salary >= 75000)  AS avg_high_salary
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY department;
````

````sql
-- COLLECT_LIST / COLLECT_SET: Aggregate rows into arrays
-- COLLECT_LIST keeps duplicates; COLLECT_SET removes them

SELECT 
    department,
    COLLECT_LIST(name)           AS all_names,
    COLLECT_SET(department)      AS unique_depts,
    ARRAY_SIZE(COLLECT_LIST(name)) AS team_size
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY department;

-- LISTAGG / STRING_AGG: Concatenate strings with a separator
SELECT 
    department,
    LISTAGG(name, ', ') WITHIN GROUP (ORDER BY name) AS team_members
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY department;
````

````sql
-- Gap and Island Analysis: Identify consecutive sequences (islands) and gaps
-- Classic SQL pattern using window functions

-- Scenario: Track employee login days; find consecutive streaks
CREATE OR REPLACE TEMP VIEW logins AS
SELECT * FROM VALUES
  (1, DATE '2026-01-01'),
  (1, DATE '2026-01-02'),
  (1, DATE '2026-01-03'),
  (1, DATE '2026-01-06'),  -- gap (Jan 4-5 missing)
  (1, DATE '2026-01-07'),
  (1, DATE '2026-01-08'),
  (2, DATE '2026-01-01'),
  (2, DATE '2026-01-03')   -- gap (Jan 2 missing)
AS t(employee_id, login_date);

-- Step 1: Find consecutive day groups using ROW_NUMBER difference method
WITH login_gaps AS (
  SELECT 
      employee_id,
      login_date,
      DATEDIFF(login_date, 
          DATE_ADD(MIN(login_date) OVER (PARTITION BY employee_id), 
                   ROW_NUMBER() OVER (PARTITION BY employee_id ORDER BY login_date) - 1)
      ) AS day_group
  FROM logins
),
islands AS (
  -- Step 2: Group by the consecutive day identifier
  SELECT 
      employee_id,
      day_group,
      MIN(login_date) AS streak_start,
      MAX(login_date) AS streak_end,
      COUNT(*)        AS streak_length
  FROM login_gaps
  GROUP BY employee_id, day_group
)
-- Step 3: Show the streaks
SELECT 
    employee_id,
    streak_start,
    streak_end,
    streak_length,
    CASE WHEN streak_length > 1 THEN 'Streak' ELSE 'Single day' END AS streak_type
FROM islands
ORDER BY employee_id, streak_start;
````

````sql
-- Map and Struct Types: Work with nested data structures

-- STRUCT: A group of named fields (like a row within a column)
CREATE OR REPLACE TEMP VIEW employee_structs AS
SELECT * FROM VALUES
  (1, 'John',  STRUCT('Python', 'SQL', 3)),
  (2, 'Mary',  STRUCT('Excel', 'PowerBI', 2)),
  (3, 'David', STRUCT('Spark', 'Scala', 5))
AS t(id, name, skills);

-- Access struct fields with dot notation
SELECT 
    id, 
    name,
    skills.col1 AS primary_skill,
    skills.col2 AS secondary_skill,
    skills.col3 AS years_experience
FROM employee_structs
ORDER BY id;

-- MAP: Key-value pairs
CREATE OR REPLACE TEMP VIEW employee_maps AS
SELECT * FROM VALUES
  (1, MAP('city', 'Seattle', 'role', 'Engineer')),
  (2, MAP('city', 'Austin', 'role', 'Analyst')),
  (3, MAP('city', 'New York', 'role', 'Manager'))
AS t(id, attributes);

-- Access map values by key
SELECT 
    id,
    attributes['city'] AS city,
    attributes['role'] AS role,
    MAP_KEYS(attributes)   AS all_keys,
    MAP_VALUES(attributes) AS all_values
FROM employee_maps
ORDER BY id;
````

### Exercise 14: Department View with Ranks (Views + Window Functions)

Create a **temp view** that ranks employees within their department by salary, then query it to find departments where the **salary gap between rank 1 and rank 2** is more than 5K.

**Requirements:**
- Step 1: Create a temp view `dept_ranked` with `employee_id`, `name`, `department`, `salary`, `salary_rank` (using `RANK()` partitioned by department)
- Step 2: Query the view to find departments where the gap between rank 1 and rank 2 exceeds 5,000

### Exercise 15: CALCULATE — Modified Filter Context

**DAX:** `CALCULATE(AVERAGE(Salary), Department = "IT")`

Write a SQL query that shows each employee's salary, the **company average salary**, and the **IT department average salary** side by side — all in one row per employee.

- `it_avg` = average salary for IT department only (same value for every row — like DAX CALCULATE overrides the filter context)
- The IT average should appear even for non-IT employees
- Order by `salary` descending

<details><summary>Hint</summary>Use two scalar subqueries in the SELECT — one with no department filter, one with `WHERE department = 'IT'`</details>

````sql
-- Solution: Exercise 15 — CALCULATE (Modified Filter Context)
-- DAX: CALCULATE(AVERAGE(Salary), Department = "IT")
-- SQL: Two scalar subqueries — one unfiltered (ALL), one filtered to IT

SELECT 
    name,
    department,
    salary,
    (SELECT ROUND(AVG(salary)) FROM employees WHERE department IS NOT NULL) AS company_avg,
    (SELECT ROUND(AVG(salary)) FROM employees WHERE department = 'IT')  AS it_avg
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;
````

### Exercise 19: SAMEPERIODLASTYEAR — Year-over-Year Comparison

**DAX:** `CALCULATE(SUM(Sales), SAMEPERIODLASTYEAR(Date))`

Write a query that compares **monthly sales** for each month in 2025 against the **same month in 2024**.

**Requirements:**

- Return `month`, `sales_2024`, `sales_2025`, `yoy_diff`, `yoy_pct_growth`
- `yoy_diff` = `sales_2025 - sales_2024`
- `yoy_pct_growth` = `ROUND((sales_2025 - sales_2024) * 100.0 / sales_2024, 1)`
- Self-join the table on `month` to get both years side by side
- Order by `month`

<details><summary>Hint</summary>Self-join `monthly_sales a JOIN monthly_sales b ON a.month = b.month AND a.year = 2024 AND b.year = 2025`</details>

````sql
-- Solution: Exercise 19 — SAMEPERIODLASTYEAR (Year-over-Year Comparison)
-- DAX: CALCULATE(SUM(Sales), SAMEPERIODLASTYEAR(Date))
-- SQL: Self-join on month with year offset

CREATE OR REPLACE TEMP VIEW monthly_sales AS
SELECT * FROM VALUES
  (2024, 1, 100), (2024, 2, 110), (2024, 3, 120), (2024, 4, 130),
  (2025, 1, 120), (2025, 2, 125), (2025, 3, 140), (2025, 4, 145)
AS t(year, month, sales_amount);

SELECT 
    b.month,
    a.sales_amount AS sales_2024,
    b.sales_amount AS sales_2025,
    b.sales_amount - a.sales_amount AS yoy_diff,
    ROUND((b.sales_amount - a.sales_amount) * 100.0 / a.sales_amount, 1) AS yoy_pct_growth
FROM monthly_sales a
JOIN monthly_sales b ON a.month = b.month AND a.year = 2024 AND b.year = 2025
ORDER BY b.month;
````

### Exercise 20: TOTALYTD — Year-to-Date Cumulative

**DAX:** `TOTALYTD(SUM(Sales), Date)`

Write a query that shows **monthly sales** and the **year-to-date cumulative total** for 2025.

- Use the `monthly_sales` temp view from Exercise 19

- `ytd_sales` = running sum of sales from month 1 through the current month (within 2025)
- Use `SUM() OVER (ORDER BY month)` for the running total
- Order by `month`

### Exercise 21: DIVIDE — Safe Division

**DAX:** `DIVIDE(numerator, denominator)` returns blank on divide-by-zero

Write a query that calculates each employee's salary as a **percentage of their department's total salary**, handling the case where the department total might be zero.

**Requirements:**
- Return `name`, `department`, `salary`, `dept_total`, `pct_of_dept`
- `pct_of_dept` = `salary / dept_total * 100` (rounded to 1 decimal)
- Use `CASE WHEN` to return `NULL` if `dept_total` is 0 (like DAX `DIVIDE`)
- Use a CTE for department totals
- Exclude employees with no department
- Order by `department`, `salary` descending

````sql
-- Solution: Exercise 21 — DIVIDE (Safe Division)
-- DAX: DIVIDE(numerator, denominator) — returns blank on divide-by-zero
-- SQL: CASE WHEN denominator = 0 THEN NULL ELSE numerator/denominator END

WITH dept_totals AS (
    SELECT department, SUM(salary) AS dept_total
    FROM employees
    WHERE department IS NOT NULL
    GROUP BY department
)
SELECT 
    e.name,
    e.department,
    e.salary,
    d.dept_total,
    CASE 
        WHEN d.dept_total = 0 THEN NULL
        ELSE ROUND(e.salary * 100.0 / d.dept_total, 1)
    END AS pct_of_dept
FROM employees e
JOIN dept_totals d ON e.department = d.department
WHERE e.department IS NOT NULL
ORDER BY e.department, e.salary DESC;
````

### Exercise 22: SUMX — Iterator with CALCULATE Context

**DAX:** `SUMX(Sales, Quantity * UnitPrice)` — iterates row by row, computes expression, then sums

Write a query that calculates the **total revenue** from a sales table where each row has `quantity` and `unit_price`, then shows each product's **contribution to total revenue**.

**Requirements:**
- Create a temp view `product_sales` with `product`, `quantity`, `unit_price`
- Return `product`, `quantity`, `unit_price`, `line_total`, `pct_of_revenue`
- `line_total` = `quantity * unit_price` (like DAX SUMX computing per-row)
- `pct_of_revenue` = each product's `line_total` as % of grand total
- Use a CTE for the per-row calculation (like DAX iterator step)
- Order by `line_total` descending

````sql
-- Solution: Exercise 22 — SUMX (Iterator with CALCULATE Context)
-- DAX: SUMX(Sales, Quantity * UnitPrice) — row-by-row compute then sum
-- SQL: CTE computes per-row expression, then aggregate

CREATE OR REPLACE TEMP VIEW product_sales AS
SELECT * FROM VALUES
  ('Widget',  100, 25),
  ('Gadget',  50,  60),
  ('Gizmo',   200, 15),
  ('Doohickey', 30, 90)
AS t(product, quantity, unit_price);

WITH line_items AS (
    -- Iterator step: compute per row (like DAX SUMX first pass)
    SELECT 
        product,
        quantity,
        unit_price,
        quantity * unit_price AS line_total
    FROM product_sales
),
grand_total AS (
    SELECT SUM(line_total) AS total_revenue
    FROM line_items
)
SELECT 
    l.product,
    l.quantity,
    l.unit_price,
    l.line_total,
    ROUND(l.line_total * 100.0 / g.total_revenue, 1) AS pct_of_revenue
FROM line_items l
CROSS JOIN grand_total g
ORDER BY l.line_total DESC;
````

### Exercise 29: SWITCH — Multi-Way Conditional Classification

**DAX:** `SWITCH(TRUE(), Salary >= 90000, "Top", Salary >= 80000, "Senior", Salary >= 70000, "Mid", "Entry")`

SWITCH evaluates conditions in order and returns the first match — identical to SQL `CASE WHEN`. Write a query that classifies employees into 4 salary tiers and counts how many are in each.

**Requirements:**
- Step 1: Create a temp view `employee_tiers` with `name`, `department`, `salary`, `tier`
- Tier logic: ≥ 90K = `'Top'`, ≥ 80K = `'Senior'`, ≥ 70K = `'Mid'`, else = `'Entry'`
- Step 2: Query the view to show `tier`, `emp_count`, `avg_salary`, `min_salary`, `max_salary`

````sql
-- Solution: Exercise 29 — SWITCH (Multi-Way Conditional Classification)
-- DAX: SWITCH(TRUE(), Salary >= 90000, "Top", Salary >= 80000, "Senior", ...)
-- SQL: CASE WHEN — evaluated in order, first match wins

-- Step 1: Create temp view with salary tiers
CREATE OR REPLACE TEMP VIEW employee_tiers AS
SELECT 
    name,
    department,
    salary,
    CASE 
        WHEN salary >= 90000 THEN 'Top'
        WHEN salary >= 80000 THEN 'Senior'
        WHEN salary >= 70000 THEN 'Mid'
        ELSE 'Entry'
    END AS tier
FROM employees
WHERE department IS NOT NULL;

-- Step 2: Aggregate by tier
SELECT 
    tier,
    COUNT(*)        AS emp_count,
    ROUND(AVG(salary)) AS avg_salary,
    MIN(salary)    AS min_salary,
    MAX(salary)    AS max_salary
FROM employee_tiers
GROUP BY tier
ORDER BY avg_salary DESC;
````

### Exercise 30: RELATEDTABLE — Count Related Rows

**DAX:** `COUNTROWS(RELATEDTABLE(Employees))` from the Department table

<details><summary>Hint</summary>Use a CTE to rank departments within each tier by emp_count. Then in the outer query, check if multiple departments share rank 1.</details>

````sql
-- Solution: Exercise 31 — SELECTEDVALUE (Get Single Value or Default)
-- DAX: SELECTEDVALUE(Department, "No selection")
-- SQL: CASE WHEN COUNT(DISTINCT ...) = 1 THEN MAX(...) ELSE 'default' END

CREATE OR REPLACE TEMP VIEW employee_tiers AS
SELECT 
    name, department, salary,
    CASE 
        WHEN salary >= 90000 THEN 'Top'
        WHEN salary >= 80000 THEN 'Senior'
        WHEN salary >= 70000 THEN 'Mid'
        ELSE 'Entry'
    END AS tier
FROM employees
WHERE department IS NOT NULL;

WITH dept_counts AS (
    SELECT 
        tier,
        department,
        COUNT(*) AS emp_count,
        RANK() OVER (PARTITION BY tier ORDER BY COUNT(*) DESC) AS rnk
    FROM employee_tiers
    GROUP BY tier, department
)
SELECT 
    tier,
    CASE 
        WHEN COUNT(*) OVER (PARTITION BY tier) > 1 THEN 'Multiple'
        ELSE MAX(department) 
    END AS top_department
FROM dept_counts
WHERE rnk = 1
GROUP BY tier, department
QUALIFY ROW_NUMBER() OVER (PARTITION BY tier ORDER BY department) = 1
ORDER BY tier;
````

### Exercise 38: VAR / RETURN — Variable Declarations for Complex Measures

**DAX:**
```

RETURN (dept_avg - company_avg) / company_avg
```

DAX `VAR` defines intermediate values. In SQL, this maps to **CTEs** — each VAR becomes a CTE. Write a query that computes each department's **deviation from the company average** as a percentage.

**Requirements:**
- Return `department`, `dept_avg`, `company_avg`, `deviation_pct`
- `deviation_pct` = `(dept_avg - company_avg) / company_avg * 100` (rounded to 1 decimal)
- Use CTEs to define intermediate values (like DAX VARs):
  - CTE 1: `dept_avg` per department
  - CTE 2: `company_avg` (single value)
  - Final SELECT: combine them
- Exclude employees with no department
- Order by `deviation_pct` descending

````sql
-- Solution: Exercise 38 — VAR / RETURN (Variable Declarations)
-- DAX: VAR dept_avg = AVERAGE(Salary); VAR company_avg = CALCULATE(...); RETURN ...
-- SQL: Each VAR becomes a CTE — intermediate values defined step by step

WITH dept_avg AS (
    -- VAR 1: per-department average
    SELECT 
        department,
        ROUND(AVG(salary)) AS dept_avg_salary
    FROM employees
    WHERE department IS NOT NULL
    GROUP BY department
),
company_avg AS (
    -- VAR 2: company-wide average (single value)
    SELECT ROUND(AVG(salary)) AS company_avg_salary
    FROM employees
    WHERE department IS NOT NULL
)
-- RETURN: combine VARs into final calculation
SELECT 
    d.department,
    d.dept_avg_salary AS dept_avg,
    c.company_avg_salary AS company_avg,
    ROUND((d.dept_avg_salary - c.company_avg_salary) * 100.0 / c.company_avg_salary, 1) AS deviation_pct
FROM dept_avg d
CROSS JOIN company_avg c
ORDER BY deviation_pct DESC;
````

````sql
-- Solution: Exercise 42 — TREATAS (Apply Table Expression as Filter)
-- DAX: CALCULATE(COUNTROWS(Employees), TREATAS({"IT", "Finance"}, Department))
-- SQL: WHERE column IN ('value1', 'value2') — TREATAS maps to IN with a value list

SELECT 
    'IT, Finance' AS filtered_depts,
    (SELECT COUNT(*) FROM employees WHERE department IN ('IT', 'Finance')) AS filtered_count,
    (SELECT COUNT(*) FROM employees WHERE department IS NOT NULL)        AS total_count,
    ROUND(
        (SELECT COUNT(*) FROM employees WHERE department IN ('IT', 'Finance')) * 100.0 /
        (SELECT COUNT(*) FROM employees WHERE department IS NOT NULL), 1
    ) AS pct_of_total;
````

### Exercise 43: ISFILTERED / HASONEVALUE — Detect Filter Context

ISFILTERED checks if a column has a filter applied. HASONEVALUE checks if exactly one value exists in context. In SQL, use `CASE WHEN` with `COUNT(DISTINCT ...)`.

Write a query that for each salary tier, reports whether the department is filtered (single department) or has multiple departments, and shows the department name if exactly one.

**Requirements:**
- Return `tier`, `is_filtered`, `department_value`

- `department_value` = department name if single, else `'N/A'`
- Use `COUNT(DISTINCT ...)` in a `CASE WHEN` (like DAX HASONEVALUE)
- Use the `employee_tiers` temp view from Exercise 29
- Order by `tier`

````sql
-- Solution: Exercise 43 — ISFILTERED / HASONEVALUE (Detect Filter Context)
-- DAX: IF(ISFILTERED(Dept), ...) and IF(HASONEVALUE(Dept), VALUES(Dept), "Multiple")
-- SQL: CASE WHEN COUNT(DISTINCT ...) = 1 THEN ... ELSE ... END

CREATE OR REPLACE TEMP VIEW employee_tiers AS
SELECT 
    name, department, salary,
    CASE 
        WHEN salary >= 90000 THEN 'Top'
        WHEN salary >= 80000 THEN 'Senior'
        WHEN salary >= 70000 THEN 'Mid'
        ELSE 'Entry'
    END AS tier
FROM employees
WHERE department IS NOT NULL;

SELECT 
    tier,
    CASE 
        WHEN COUNT(DISTINCT department) = 1 THEN 'Single dept'
        ELSE 'Multiple depts'
    END AS is_filtered,
    CASE 
        WHEN COUNT(DISTINCT department) = 1 THEN MAX(department)
        ELSE 'N/A'
    END AS department_value
FROM employee_tiers
GROUP BY tier
ORDER BY tier;
````

### Exercise 44: PARALLELPERIOD — Shift Entire Period

**DAX:** `CALCULATE(SUM(Sales), PARALLELPERIOD(Date, -1, YEAR))`

PARALLELPERIOD shifts the entire period by N intervals (e.g., previous year). In SQL, this is a `LAG` on an aggregated result, or a self-join with a year offset.

Write a query that for each month, shows the current year's sales, the **previous year's same-month sales**, and the difference.

**Requirements:**
- Use the `monthly_sales` temp view from Exercise 19
- Return `month`, `sales_2024`, `sales_2025`, `period_diff`
- `period_diff` = `sales_2025 - sales_2024`
- Use `LAG` window function on the aggregated data (like PARALLELPERIOD shifting back 1 year = 4 rows)
- Order by `year`, `month`
- Show all 8 rows including 2024 (with NULL for prior year)

````sql
-- Solution: Exercise 44 — PARALLELPERIOD (Shift Entire Period)
-- DAX: CALCULATE(SUM(Sales), PARALLELPERIOD(Date, -1, YEAR))
-- SQL: LAG(sales, 4) OVER (ORDER BY year, month) — shift by 4 rows (1 year of monthly data)

SELECT 
    year,
    month,
    sales_amount AS current_sales,
    LAG(sales_amount, 4) OVER (ORDER BY year, month) AS parallel_period_sales,
    sales_amount - LAG(sales_amount, 4) OVER (ORDER BY year, month) AS period_diff
FROM monthly_sales
ORDER BY year, month;
````

### Exercise 45: MAXX / MINX — Iterator Max/Min

**DAX:** `MAXX(Employees, Salary)` per department and `MINX(Employees, Salary)` per department

MAXX/MINX iterate over a table and return the max/min of an expression. In SQL, this is `MAX()` / `MIN()` with `GROUP BY`, or a window function to show the max/min alongside each row.

Write a query that for each department, shows the **highest and lowest salary**, plus the **spread** (max - min).

**Requirements:**

- Use `MAX()` and `MIN()` with `GROUP BY` (like DAX MAXX/MINX over a filtered table)

- Order by `salary_spread` descending

````sql
-- Solution: Exercise 45 — MAXX / MINX (Iterator Max/Min)
-- DAX: MAXX(Employees, Salary) per department and MINX(Employees, Salary) per department
-- SQL: MAX() / MIN() with GROUP BY

SELECT 
    department,
    MAX(salary) AS max_salary,
    MIN(salary) AS min_salary,
    MAX(salary) - MIN(salary) AS salary_spread
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY salary_spread DESC;
````

````sql
-- SEMI JOIN / ANTI JOIN: Return rows from one table based on existence in another
-- SEMI JOIN: Returns left rows that have a match (like INNER JOIN but no right columns, no duplicates)
-- ANTI JOIN: Returns left rows that have NO match (like LEFT JOIN ... WHERE right IS NULL, cleaner)
-- In Databricks SQL, use EXISTS / NOT EXISTS (most readable) or IN / NOT IN

-- SEMI JOIN: Employees who belong to a registered department
SELECT employee_id, name, department, salary
FROM employees e
WHERE EXISTS (
    SELECT 1 FROM departments d WHERE d.department = e.department
)
ORDER BY employee_id;

-- ANTI JOIN: Employees NOT in any registered department
SELECT employee_id, name, salary
FROM employees e
WHERE NOT EXISTS (
    SELECT 1 FROM departments d WHERE d.department = e.department
)
ORDER BY employee_id;

-- Equivalent with IN / NOT IN (same result, but EXISTS is preferred for clarity)
SELECT name, department FROM employees
WHERE department IN (SELECT department FROM departments)
ORDER BY employee_id;

SELECT name, salary FROM employees
WHERE department NOT IN (SELECT department FROM departments)
   OR department IS NULL
ORDER BY employee_id;
````

````sql
-- USING & NATURAL JOIN: Simplified join syntax
-- USING: Join on a shared column name (both tables must have the same column name)
-- NATURAL JOIN: Automatically joins on ALL columns with matching names

-- Create temp views with a shared 'department' column name for USING demo
CREATE OR REPLACE TEMP VIEW emp_with_dept AS
SELECT employee_id, name, department, salary FROM employees WHERE department IS NOT NULL;

CREATE OR REPLACE TEMP VIEW dept_mgr AS
SELECT department, manager FROM departments;

-- USING clause: Join on the shared 'department' column (no need for ON clause)
-- No need to qualify 'department' — it appears once in the output
SELECT employee_id, name, department, salary, manager
FROM emp_with_dept e
JOIN dept_mgr d USING (department)
ORDER BY employee_id;

-- NATURAL JOIN: Automatically joins on all matching column names
-- Here, 'department' is the only shared column, so it joins on that
SELECT employee_id, name, salary, manager
FROM emp_with_dept
NATURAL JOIN dept_mgr
ORDER BY employee_id;

-- Note: NATURAL JOIN is risky in production — if both tables add a column
-- with the same name later, it silently changes the join condition. Prefer USING.
````

````sql
-- Advanced Array Functions: Sort, combine, filter, and transform arrays

CREATE OR REPLACE TEMP VIEW array_demo AS
SELECT * FROM VALUES
  ('Alice', ARRAY(95, 87, 92, 78)),
  ('Bob',   ARRAY(70, 65, 80)),
  ('Carol', ARRAY(100, 85))
AS t(name, scores);

SELECT * FROM array_demo ORDER BY name;

-- ARRAY_SORT: Sort array elements ascending
SELECT name, scores, ARRAY_SORT(scores) AS sorted_scores
FROM array_demo
ORDER BY name;

-- ARRAY_MIN / ARRAY_MAX: Get min/max from array
SELECT 
    name,
    ARRAY_MIN(scores) AS min_score,
    ARRAY_MAX(scores) AS max_score,
    ARRAY_MAX(scores) - ARRAY_MIN(scores) AS score_range
FROM array_demo
ORDER BY name;

-- ARRAY_CONTAINS: Check if a value exists in an array
SELECT name, scores, ARRAY_CONTAINS(scores, 90) AS has_90_plus
FROM array_demo
ORDER BY name;

-- ARRAY_REMOVE: Remove all occurrences of a value
SELECT name, ARRAY_REMOVE(scores, 100) AS scores_no_perfect
FROM array_demo
ORDER BY name;

-- TRANSFORM: Apply an expression to each element (like map in Python)
-- Add 5 bonus points to each score
SELECT name, TRANSFORM(scores, x -> x + 5) AS boosted_scores
FROM array_demo
ORDER BY name;

-- FILTER: Keep only elements matching a condition
SELECT name, FILTER(scores, x -> x >= 85) AS high_scores
FROM array_demo
ORDER BY name;

-- AGGREGATE: Reduce array to a single value (like reduce in Python)
-- AGGREGATE(array, initial_value, (acc, x) -> expr, final_expr)
-- Sum of scores (identity is 0)
SELECT 
    name, 
    scores,
    AGGREGATE(scores, 0, (acc, x) -> acc + x) AS total_score,
    AGGREGATE(scores, 0, (acc, x) -> acc + x) / ARRAY_SIZE(scores) AS avg_score
FROM array_demo
ORDER BY name;

-- ARRAY_UNION: Combine arrays (deduplicates)
SELECT ARRAY_UNION(ARRAY(1, 2, 3), ARRAY(3, 4, 5)) AS union_result;

-- ARRAY_INTERSECT: Common elements only
SELECT ARRAY_INTERSECT(ARRAY(1, 2, 3, 4), ARRAY(3, 4, 5, 6)) AS intersect_result;

-- ARRAY_EXCEPT: Elements in first but not second
SELECT ARRAY_EXCEPT(ARRAY(1, 2, 3, 4), ARRAY(3, 4, 5)) AS except_result;

-- ARRAY_DISTINCT: Remove duplicates from array
SELECT ARRAY_DISTINCT(ARRAY(1, 1, 2, 3, 3, 3)) AS distinct_result;
````

````sql
-- TRY_* Safe Functions: Graceful error handling without crashing the query
-- TRY_CAST: Convert type, return NULL on failure (instead of error)
-- TRY_DIVIDE: Divide, return NULL on divide-by-zero (instead of error)
-- NULLIF: Return NULL if two values are equal (prevents divide-by-zero)
-- NVL / NVL2: Oracle-style NULL coalescing (like COALESCE)

-- TRY_CAST: Convert strings to numbers, return NULL for invalid values
-- The query below intentionally uses CAST (which errors) — commented out for safety:
-- SELECT col_val, TRY_CAST(col_val AS INT) AS try_int, CAST(col_val AS INT) AS cast_int
--   FROM VALUES ('123'), ('45.6'), ('abc'), ('789') AS t(col_val);
-- CAST('abc' AS INT) would crash the query, but TRY_CAST returns NULL instead.

-- Safe version: only TRY_CAST (never errors)
SELECT 
    col_val,
    TRY_CAST(col_val AS INT)      AS try_int,
    TRY_CAST(col_val AS DOUBLE)   AS try_double
FROM VALUES 
  ('123'), 
  ('45.6'), 
  ('abc'), 
  ('789')
AS t(col_val);

-- TRY_DIVIDE: Safe division, returns NULL on divide-by-zero
-- normal_result below would crash on row 2 (50/0) — that's why TRY_DIVIDE exists
SELECT 
    numerator,
    denominator,
    TRY_DIVIDE(numerator, denominator) AS safe_result
    -- numerator / denominator AS normal_result  -- would error on zero
FROM VALUES 
  (100, 5), 
  (50, 0), 
  (75, 3)
AS t(numerator, denominator);

-- NULLIF: Return NULL if two values are equal (useful for safe division)
-- Equivalent to: CASE WHEN a = b THEN NULL ELSE a END
SELECT 
    salary,
    NULLIF(salary, 80000) AS null_if_80k,  -- Returns NULL when salary = 80000
    COALESCE(NULLIF(salary, 80000), -1)   AS coalesced     -- Fall back to -1
FROM employees
ORDER BY salary DESC;

-- NVL: Return first non-NULL value (same as COALESCE with 2 args)
SELECT 
    name,
    department,
    NVL(department, 'Unassigned') AS dept_nvl,
    NVL2(department, 'Has dept', 'No dept') AS dept_status  -- NVL2: if NOT NULL → arg2, else arg3
FROM employees
ORDER BY employee_id;
````

````sql
-- EXTRACT / DATE_PART: Pull individual components from dates and timestamps
-- EXTRACT(field FROM source): SQL-standard syntax
-- DATE_PART(field, source): Databricks function form (same result)

SELECT 
    CURRENT_DATE()                              AS today,
    EXTRACT(YEAR   FROM CURRENT_DATE())         AS yr,
    EXTRACT(MONTH  FROM CURRENT_DATE())         AS mon,
    EXTRACT(DAY    FROM CURRENT_DATE())         AS dy,
    EXTRACT(QUARTER FROM CURRENT_DATE())        AS qtr,
    EXTRACT(DOW    FROM CURRENT_DATE())         AS day_of_week,   -- 0=Sunday
    EXTRACT(DOY    FROM CURRENT_DATE())         AS day_of_year;

-- DATE_PART: same logic, different syntax (more composable in expressions)
SELECT 
    CURRENT_DATE()                          AS today,
    DATE_PART('year',    CURRENT_DATE())    AS yr,
    DATE_PART('month',   CURRENT_DATE())    AS mon,
    DATE_PART('day',      CURRENT_DATE())    AS dy,
    DATE_PART('quarter',  CURRENT_DATE())    AS qtr;

-- Practical: group sales by extracted month and year from the monthly_sales view
SELECT 
    year,
    month,
    sales_amount,
    EXTRACT(QUARTER FROM MAKE_DATE(year, month, 1)) AS sales_quarter,
    CASE EXTRACT(MONTH FROM MAKE_DATE(year, month, 1))
        WHEN 1 THEN 'Q1-Start'
        WHEN 4 THEN 'Q2-Start'
        WHEN 7 THEN 'Q3-Start'
        WHEN 10 THEN 'Q4-Start'
        ELSE 'Mid-quarter'
    END AS quarter_position
FROM monthly_sales
ORDER BY year, month;
````

````sql
-- GREATEST / LEAST: Return the max/min across multiple COLUMNS (not rows)
-- Useful when you need the best/worst value per row from several columns

-- GREATEST: largest value across columns (ignores NULLs if any arg is non-NULL)
SELECT 
    name,
    salary,
    GREATEST(salary, 80000)        AS at_least_80k,
    LEAST(salary, 80000)           AS at_most_80k,
    GREATEST(salary - 70000, 0)    AS salary_above_70k_floor  -- clamp to 0
FROM employees
ORDER BY salary DESC;

-- GREATEST with multiple columns from different sources
CREATE OR REPLACE TEMP VIEW emp_review AS
SELECT * FROM VALUES
  (1, 'John',   80000, 85000, 78000),
  (2, 'Mary',   70000, 72000, 69000),
  (3, 'David',  90000, 92000, 88000),
  (4, 'Lisa',   85000, 80000, 87000)
AS t(emp_id, name, q1_salary, q2_salary, q3_salary);

-- Find the best and worst quarter per employee
SELECT 
    name,
    q1_salary,
    q2_salary,
    q3_salary,
    GREATEST(q1_salary, q2_salary, q3_salary) AS best_quarter,
    LEAST(q1_salary, q2_salary, q3_salary)    AS worst_quarter,
    GREATEST(q1_salary, q2_salary, q3_salary) 
      - LEAST(q1_salary, q2_salary, q3_salary) AS salary_swing
FROM emp_review
ORDER BY salary_swing DESC;
````

````sql
-- IF / IIF / DECODE: Inline conditional functions (shorter than CASE WHEN)

-- IF(condition, true_value, false_value): Spark SQL shorthand for CASE WHEN
SELECT 
    name,
    salary,
    IF(salary >= 80000, 'High earner', 'Standard') AS earner_type
FROM employees
ORDER BY salary DESC;

-- IIF(condition, true_val, false_val): SQL Server-style inline if
-- Note: IIF is NOT built-in in Databricks SQL; use IF instead (same result)
SELECT 
    name,
    department,
    IF(department IS NULL, 'Unassigned', department) AS dept_resolved
FROM employees
ORDER BY employee_id;

-- DECODE(value, key1, result1, key2, result2, ..., default): Oracle-style multi-way switch
-- Like a compact CASE WHEN value = key THEN result
SELECT 
    name,
    department,
    salary,
    DECODE(department,
        'IT',      'Tech',
        'HR',      'People Ops',
        'Finance', 'Accounting',
        'Sales',   'Revenue',
        'Other'
    ) AS dept_label,
    DECODE(TRUE,
        salary >= 85000, 'Top',
        salary >= 70000, 'Mid',
        'Entry'
    ) AS salary_band
FROM employees
ORDER BY salary DESC;

-- Chaining IF for multi-condition (less readable than CASE, but concise)
SELECT 
    name,
    salary,
    IF(salary >= 90000, 'Elite',
       IF(salary >= 80000, 'Senior',
          IF(salary >= 70000, 'Mid', 'Entry'))) AS tier
FROM employees
ORDER BY salary DESC;
````

````sql
-- NTILE / PERCENT_RANK / CUME_DIST: Advanced distribution window functions
-- NTILE(N): Divide ordered rows into N equal-sized buckets (1 to N)
-- PERCENT_RANK(): Relative rank as a percentage (0.0 to 1.0)
-- CUME_DIST(): Cumulative distribution — fraction of rows at or below the current row

-- NTILE: Divide employees into 3 salary groups (top third, middle third, bottom third)
SELECT 
    name,
    department,
    salary,
    NTILE(3) OVER (ORDER BY salary DESC) AS salary_tertile
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- NTILE(4) = quartiles
SELECT 
    name,
    salary,
    NTILE(4) OVER (ORDER BY salary DESC) AS quartile
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- PERCENT_RANK: Where each row sits relative to the group (0.0 = top, 1.0 = bottom)
SELECT 
    name,
    salary,
    PERCENT_RANK() OVER (ORDER BY salary DESC) AS pct_rank,
    CUME_DIST()   OVER (ORDER BY salary DESC) AS cume_dist
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- Per-department NTILE: rank within each department
SELECT 
    department,
    name,
    salary,
    NTILE(2) OVER (PARTITION BY department ORDER BY salary DESC) AS dept_half
FROM employees
WHERE department IS NOT NULL
ORDER BY department, salary DESC;
````

````sql
-- INTERVAL Literals: Express time spans for date arithmetic
-- INTERVAL 'N unit': A typed duration (not a point in time)
-- Useful for adding/subtracting time from dates and timestamps

SELECT 
    DATE '2026-10-06'                              AS start_date,
    DATE '2026-10-06' + INTERVAL '7' DAY           AS plus_7_days,
    DATE '2026-10-06' - INTERVAL '1' MONTH          AS minus_1_month,
    DATE '2026-10-06' + INTERVAL '3' MONTH          AS plus_3_months,
    DATE '2026-10-06' + INTERVAL '1' YEAR          AS plus_1_year,
    DATE '2026-10-06' + INTERVAL '90' DAY           AS plus_90_days;

-- INTERVAL with TIMESTAMP: add hours, minutes, seconds
SELECT 
    TIMESTAMP '2026-10-06 14:30:00'                          AS start_ts,
    TIMESTAMP '2026-10-06 14:30:00' + INTERVAL '2' HOUR      AS plus_2_hours,
    TIMESTAMP '2026-10-06 14:30:00' + INTERVAL '45' MINUTE   AS plus_45_min,
    TIMESTAMP '2026-10-06 14:30:00' + INTERVAL '1 2' DAY TO HOUR AS plus_1day_2hr;

-- Practical: Calculate tenure from hire date using INTERVAL
CREATE OR REPLACE TEMP VIEW tenure_demo AS
SELECT * FROM VALUES
  ('John',  DATE '2024-03-15'),
  ('Mary',  DATE '2025-01-10'),
  ('David', DATE '2023-06-01')
AS t(name, hire_date);

SELECT 
    name,
    hire_date,
    CURRENT_DATE() - hire_date                       AS days_employed,
    (CURRENT_DATE() - hire_date) / 30                 AS approx_months,
    hire_date + INTERVAL '1' YEAR                    AS first_anniversary,
    hire_date + INTERVAL '90' DAY                     AS first_quarter_review
FROM tenure_demo
ORDER BY hire_date;
````

````sql
-- Stored Procedures: Reusable multi-statement SQL blocks in Unity Catalog
-- CREATE PROCEDURE defines a callable routine with parameters
--
-- NOTE: Stored procedures require a SQL warehouse (not serverless interactive compute).
-- The commands below are commented out for reference; run them on a SQL warehouse.

-- CREATE OR REPLACE PROCEDURE sql_dept_summary(IN dept_name STRING)
-- LANGUAGE SQL
-- READS SQL DATA
-- AS $$
--   BEGIN
--     SELECT 
--         name,
--         salary,
--         salary - (SELECT AVG(salary) FROM employees WHERE department = dept_name) AS diff_from_avg
--     FROM employees
--     WHERE department = dept_name
--     ORDER BY salary DESC;
--   END
-- $$;
--
-- -- Call the procedure
-- CALL sql_dept_summary('IT');
-- CALL sql_dept_summary('HR');

-- Equivalent using a SQL UDF (works on serverless — already demonstrated in topic 88)
-- CREATE OR REPLACE FUNCTION sql_dept_avg(dept STRING) RETURNS DOUBLE
-- RETURN (SELECT AVG(salary) FROM employees WHERE department = dept);

-- Use the UDF inline (same result as the procedure would return)
SELECT 
    name,
    salary,
    salary - (SELECT AVG(salary) FROM employees WHERE department = 'IT') AS diff_from_dept_avg
FROM employees
WHERE department = 'IT'
ORDER BY salary DESC;
````

````sql
-- SOUNDEX / LEVENSHTEIN: Fuzzy string matching and edit distance
-- SOUNDEX: Phonetic algorithm — words that sound similar get the same code
-- LEVENSHTEIN: Number of single-char edits (insert/delete/replace) to transform one string into another

-- SOUNDEX: Find names that sound alike (useful for deduplication)
SELECT 
    name,
    SOUNDEX(name) AS sound_code
FROM employees
WHERE department IS NOT NULL
ORDER BY name;

-- Create a matching scenario
CREATE OR REPLACE TEMP VIEW customer_names AS
SELECT * FROM VALUES
  ('Jon Smith'),    ('John Smyth'),    ('Jon Smythe'),
  ('Mary Jones'),   ('Marry Jons'),    ('Mary Johnson'),
  ('David Lee'),    ('Daivd Li'),      ('David Leigh')
AS t(name);

-- Group by SOUNDEX to find phonetically similar names
SELECT 
    SOUNDEX(name) AS sound_code,
    COLLECT_LIST(name) AS similar_names,
    COUNT(*) AS match_count
FROM customer_names
GROUP BY SOUNDEX(name)
ORDER BY match_count DESC;

-- LEVENSHTEIN: Edit distance between two strings (lower = more similar)
SELECT 
    a.name AS name1,
    b.name AS name2,
    LEVENSHTEIN(a.name, b.name) AS edit_distance
FROM customer_names a
CROSS JOIN customer_names b
WHERE a.name < b.name
  AND LEVENSHTEIN(a.name, b.name) <= 3
ORDER BY edit_distance;
````

````sql
-- WIDTH_BUCKET: Assign values to histogram buckets
-- WIDTH_BUCKET(value, min, max, num_buckets) → bucket number (1 to num_buckets+1)
-- Values below min → bucket 0, values above max → bucket num_buckets+1

-- Divide salaries into 4 equal-width buckets between 60K and 100K
SELECT 
    name,
    salary,
    WIDTH_BUCKET(salary, 60000, 100000, 4) AS salary_bucket,
    CASE WIDTH_BUCKET(salary, 60000, 100000, 4)
        WHEN 1 THEN '60K-70K'
        WHEN 2 THEN '70K-80K'
        WHEN 3 THEN '80K-90K'
        WHEN 4 THEN '90K-100K'
        ELSE 'Out of range'
    END AS bucket_label
FROM employees
ORDER BY salary DESC;

-- Histogram: count employees per salary bucket
SELECT 
    WIDTH_BUCKET(salary, 60000, 100000, 4) AS bucket_num,
    CASE WIDTH_BUCKET(salary, 60000, 100000, 4)
        WHEN 1 THEN '60K-70K'
        WHEN 2 THEN '70K-80K'
        WHEN 3 THEN '80K-90K'
        WHEN 4 THEN '90K-100K'
        ELSE 'Out of range'
    END AS salary_range,
    COUNT(*) AS emp_count
FROM employees
GROUP BY WIDTH_BUCKET(salary, 60000, 100000, 4)
ORDER BY bucket_num;

-- WIDTH_BUCKET on test scores
CREATE OR REPLACE TEMP VIEW test_scores AS
SELECT * FROM VALUES
  (85), (92), (78), (65), (95), (70), (88), (72), (60), (99), (82), (75)
AS t(score);

-- 5-grade histogram: F (<60), D (60-70), C (70-80), B (80-90), A (90-100)
SELECT 
    WIDTH_BUCKET(score, 60, 100, 4) AS grade_bucket,
    CASE WIDTH_BUCKET(score, 60, 100, 4)
        WHEN 1 THEN 'D (60-70)'
        WHEN 2 THEN 'C (70-80)'
        WHEN 3 THEN 'B (80-90)'
        WHEN 4 THEN 'A (90-100)'
        ELSE 'F (<60)'
    END AS grade,
    COUNT(*) AS student_count
FROM test_scores
GROUP BY WIDTH_BUCKET(score, 60, 100, 4)
ORDER BY grade_bucket;
````

````sql
-- Bitwise Operations: Manipulate individual bits of integers
-- &: AND, |: OR, ^: XOR, ~: NOT, <<: left shift, >>: right shift

-- Basic bitwise operations
SELECT 
    12 AS a,
    10 AS b,
    a & b  AS bit_and,    -- 12 & 10 = 8  (1000)
    a | b  AS bit_or,     -- 12 | 10 = 14 (1110)
    a ^ b  AS bit_xor,    -- 12 ^ 10 = 6  (0110)
    ~a     AS bit_not_a,  -- ~12 = -13   (two's complement)
    a << 2 AS left_shift, -- 12 << 2 = 48 (shift left 2 bits = multiply by 4)
    a >> 2 AS right_shift -- 12 >> 2 = 3  (shift right 2 bits = divide by 4)
FROM VALUES (12, 10) AS t(a, b);

-- Practical: use bitwise flags for permissions
CREATE OR REPLACE TEMP VIEW user_perms AS
SELECT * FROM VALUES
  ('Alice', 7),    -- 0b111 = read + write + execute
  ('Bob',   5),    -- 0b101 = read + execute (no write)
  ('Carol', 3),    -- 0b011 = read + write (no execute)
  ('Dan',   1)     -- 0b001 = read only
AS t(username, perm_flags);

-- Check permissions with bitwise AND
-- Bit 1 (1) = read, Bit 2 (2) = write, Bit 3 (4) = execute
SELECT 
    username,
    perm_flags,
    (perm_flags & 1) > 0 AS can_read,
    (perm_flags & 2) > 0 AS can_write,
    (perm_flags & 4) > 0 AS can_execute,
    CASE 
        WHEN perm_flags = 7 THEN 'Full access'
        WHEN (perm_flags & 2) > 0 THEN 'Read + Write'
        WHEN (perm_flags & 1) > 0 THEN 'Read only'
        ELSE 'No access'
    END AS access_level
FROM user_perms
ORDER BY perm_flags DESC;
````

````sql
-- URL Functions: Parse and manipulate URL strings

CREATE OR REPLACE TEMP VIEW urls AS
SELECT * FROM VALUES
  ('https://example.com/products/item?id=42&cat=books'),
  ('https://api.databricks.com/v2/warehouses?type=serverless&region=us-west'),
  ('https://docs.databricks.com/sql/functions.html#date')
AS t(url);

-- PARSE_URL: Extract components from a URL
SELECT 
    url,
    PARSE_URL(url, 'PROTOCOL')  AS protocol,
    PARSE_URL(url, 'HOST')      AS host,
    PARSE_URL(url, 'PATH')      AS path,
    PARSE_URL(url, 'QUERY')     AS query,
    PARSE_URL(url, 'REF')       AS fragment
FROM urls;

-- Extract specific query parameters with PARSE_URL + QUERY key
SELECT 
    url,
    PARSE_URL(url, 'QUERY', 'id')  AS id_param,
    PARSE_URL(url, 'QUERY', 'cat') AS cat_param,
    PARSE_URL(url, 'QUERY', 'type') AS type_param
FROM urls;

-- URL_ENCODE / URL_DECODE: Handle special characters in URLs
SELECT 
    raw_text,
    URL_ENCODE(raw_text) AS encoded,
    URL_DECODE(URL_ENCODE(raw_text)) AS decoded_back
FROM VALUES 
  ('hello world & foo=bar'),
  ('price: $1,000.00'),
  ('café résumé naïve')
AS t(raw_text);
````

````sql
-- Advanced String Functions: SPLIT, SPLIT_PART, OVERLAY, REGEXP_COUNT, REPEAT, SPACE

-- SPLIT: Break a string into an array by delimiter
SELECT 
    full_name,
    SPLIT(full_name, ' ') AS name_parts
FROM VALUES 
  ('John William Smith'), 
  ('Mary Jane Watson'), 
  ('David Lee')
AS t(full_name);

-- SPLIT_PART: Get the Nth element from a delimited string (1-indexed)
SELECT 
    full_name,
    SPLIT_PART(full_name, ' ', 1) AS first_name,
    SPLIT_PART(full_name, ' ', 2) AS middle_name,
    SPLIT_PART(full_name, ' ', -1) AS last_name  -- -1 = last element
FROM VALUES 
  ('John William Smith'), 
  ('Mary Jane Watson'), 
  ('David Lee')
AS t(full_name);

-- OVERLAY: Replace a substring at a specific position (like substring assignment)
SELECT 
    original,
    OVERLAY(original PLACING 'XX' FROM 1 FOR 2) AS overlay_first2,
    OVERLAY(original PLACING '***' FROM 6) AS overlay_from6
FROM VALUES 
  ('Hello World'),
  ('ABCDEFGHIJ')
AS t(original);

-- REGEXP_COUNT: Count how many times a pattern matches
SELECT 
    text,
    REGEXP_COUNT(text, '[aeiou]') AS vowel_count,
    REGEXP_COUNT(text, '\\d+') AS number_groups
FROM VALUES 
  ('Hello World 123'),
  ('abc123def456ghi789'),
  ('The quick brown fox')
AS t(text);

-- REPEAT: Repeat a string N times
-- SPACE: Generate N spaces (useful for padding)
SELECT 
    REPEAT('ab', 3)  AS repeated,
    REPEAT('-', 10)  AS dash_line,
    SPACE(5) || 'indented' AS spaced_text,
    LENGTH(SPACE(5)) AS space_len;

-- Practical: Pad a string to fixed width
SELECT 
    name,
    RPAD(name, 15, '.') || salary AS dotted_format
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;
````

````sql
-- Base Conversion: CONV, HEX, UNHEX, BIN, OCT
-- Convert numbers between bases (binary, octal, decimal, hexadecimal)

-- CONV: Convert between numeric bases (from_base, to_base)
SELECT 
    CONV('255', 10, 16)  AS dec_to_hex,    -- 'FF'
    CONV('FF', 16, 10)   AS hex_to_dec,    -- '255'
    CONV('255', 10, 2)   AS dec_to_bin,    -- '11111111'
    CONV('777', 8, 10)   AS oct_to_dec,    -- '511'
    CONV('1010', 2, 10)  AS bin_to_dec      -- '10'
FROM VALUES (1) AS t(dummy);

-- HEX / UNHEX: Hexadecimal encode/decode
SELECT 
    HEX('Hello')     AS hex_encoded,   -- '48656C6C6F'
    UNHEX('48656C6C6F') AS hex_decoded,  -- 'Hello'
    HEX(255)         AS hex_of_num;    -- 'FF'

-- BIN / OCT: Binary and octal string representation
SELECT 
    BIN(255)  AS binary_str,   -- '11111111'
    CONV('255', 10, 8) AS octal_str,  -- '377' (OCT not built-in, use CONV)
    BIN(10)   AS bin_ten;     -- '1010'

-- Practical: Generate a hex color from a number
SELECT 
    n,
    CONCAT('#', LPAD(HEX(n), 6, '0')) AS hex_color
FROM VALUES (0), (255), (65280), (16711680), (16777215) AS t(n);
````

````sql
-- NAMED_STRUCT / STACK / INLINE: Struct creation and row multiplication

-- NAMED_STRUCT: Create a struct with named fields (alternative to STRUCT())
SELECT 
    name,
    NAMED_STRUCT('dept', department, 'pay', salary) AS emp_info
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- Access struct fields with dot notation
SELECT 
    name,
    NAMED_STRUCT('dept', department, 'pay', salary) AS emp_info,
    NAMED_STRUCT('dept', department, 'pay', salary).dept AS dept_field,
    NAMED_STRUCT('dept', department, 'pay', salary).pay AS pay_field
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- STACK: Expand a set of values into multiple rows (like UNPIVOT)
-- STACK(n, val1, val2, ..., valN) → N rows with one column each
SELECT 
    dept_name,
    STACK(3, 'Q1', 100, 'Q2', 200, 'Q3', 150) AS (quarter, sales)
FROM (SELECT 'IT' AS dept_name)
ORDER BY quarter;

-- Practical: UNPIVOT quarterly data using STACK
CREATE OR REPLACE TEMP VIEW quarterly_data AS
SELECT * FROM VALUES
  ('IT',    100, 120, 140, 160),
  ('HR',     80,  90,  85,  95),
  ('Sales',  60,  70,  65,  80)
AS t(dept, q1, q2, q3, q4);

SELECT * FROM quarterly_data;

-- Unpivot using STACK
SELECT 
    dept,
    STACK(4, 'Q1', q1, 'Q2', q2, 'Q3', q3, 'Q4', q4) AS (quarter, revenue)
FROM quarterly_data
ORDER BY dept, quarter;

-- INLINE: Flatten an array of structs into rows (like EXPLODE for structs)
SELECT 
    INLINE(ARRAY(
        NAMED_STRUCT('name', 'Alice', 'score', 95),
        NAMED_STRUCT('name', 'Bob',   'score', 87),
        NAMED_STRUCT('name', 'Carol', 'score', 92)
    ))
ORDER BY score DESC;
````

````sql
-- Advanced Math Functions: POWER, SQRT, CBRT, EXP, LN, LOG, FACTORIAL, PI, E, SIGN, TRUNC
-- Beyond basic arithmetic — scientific and statistical math

SELECT 
    POWER(2, 10)    AS two_pow_10,    -- 1024
    SQRT(144)        AS sqrt_144,      -- 12
    CBRT(27)        AS cbrt_27,       -- 3
    EXP(1)          AS e_power_1,    -- 2.718281828459045
    LN(2.718281828459045) AS ln_e,   -- ~1
    LOG(10, 1000)   AS log10_1000,   -- 3
    LOG2(256)       AS log2_256,     -- 8
    FACTORIAL(5)    AS fact_5,        -- 120
    PI()            AS pi_val,        -- 3.141592653589793
    E()             AS e_val;         -- 2.718281828459045

-- SIGN: Returns -1, 0, or 1 based on the sign
-- TRUNC: Truncate to N decimal places (not round)
SELECT 
    salary,
    SIGN(salary - 75000) AS sign_vs_75k,   -- positive (1), zero (0), or negative (-1)
    FLOOR(salary * 0.10 * 100) / 100 AS floor_bonus,  -- truncate to 2 decimals via FLOOR
    ROUND(salary * 0.10, 2) AS round_bonus   -- rounded to 2 decimals
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- Practical: compound interest calculation
SELECT 
    principal,
    rate,
    years,
    ROUND(principal * POWER(1 + rate, years), 2) AS final_amount,
    ROUND(principal * POWER(1 + rate, years) - principal, 2) AS interest_earned
FROM VALUES 
  (10000, 0.05, 10),
  (50000, 0.07, 20),
  (25000, 0.03, 5)
AS t(principal, rate, years)
ORDER BY final_amount DESC;
````

````sql
-- Collection Operations: ARRAY_JOIN, SLICE, FLATTEN, ZIP, SHUFFLE, REVERSE
-- Manipulate arrays beyond basic EXPLODE/ARRAY_SORT

CREATE OR REPLACE TEMP VIEW coll_demo AS
SELECT * FROM VALUES
  ('Alice', ARRAY(1, 2, 3, 4, 5)),
  ('Bob',   ARRAY(10, 20, 30)),
  ('Carol', ARRAY(100))
AS t(name, nums);

-- ARRAY_JOIN: Concatenate array elements into a string with delimiter
SELECT name, nums, ARRAY_JOIN(nums, ', ') AS joined
FROM coll_demo ORDER BY name;

-- SLICE: Extract a sub-array (1-indexed, length)
SELECT name, nums, SLICE(nums, 1, 3) AS first_three
FROM coll_demo ORDER BY name;

-- REVERSE: Reverse array elements
SELECT name, nums, REVERSE(nums) AS reversed_nums
FROM coll_demo ORDER BY name;

-- FLATTEN: Flatten an array of arrays into a single array
SELECT FLATTEN(ARRAY(ARRAY(1, 2), ARRAY(3, 4), ARRAY(5))) AS flattened;

-- ZIP_WITH: Combine arrays element-wise with a function
-- (This is the Databricks SQL equivalent of Python's zip + map)
SELECT 
    ZIP_WITH(ARRAY(1, 2, 3), ARRAY(10, 20, 30), (x, y) -> x + y) AS summed,
    ZIP_WITH(ARRAY('a', 'b', 'c'), ARRAY(1, 2, 3), (x, y) -> CONCAT(x, CAST(y AS STRING))) AS zipped_pairs;

-- SHUFFLE: Randomly reorder array elements (may not exist in all DBR versions)
-- SELECT name, nums, SHUFFLE(nums) AS shuffled FROM coll_demo ORDER BY name;
-- Alternative: use ZIP_WITH with RAND to pair and sort
SELECT name, nums, REVERSE(nums) AS reversed_demo
FROM coll_demo ORDER BY name;

-- Combine arrays with labels using ZIP_WITH
SELECT 
    ZIP_WITH(ARRAY('x', 'y', 'z'), ARRAY(100, 200, 300), (k, v) -> CONCAT(k, '=', CAST(v AS STRING))) AS labeled_pairs;
````

````sql
-- CONCAT_WS / FORMAT_STRING: Advanced string formatting
-- CONCAT_WS: Concatenate with a separator (like CSV building)
-- FORMAT_STRING: printf-style formatted strings

-- CONCAT_WS: Join with a delimiter, skipping NULLs
SELECT 
    name,
    department,
    salary,
    CONCAT_WS(' | ', name, department, CAST(salary AS STRING)) AS formatted,
    CONCAT_WS(', ', name, department, NULL) AS with_null_skipped
FROM employees
WHERE department IS NOT NULL
ORDER BY name;

-- FORMAT_STRING: C-style printf formatting
-- %s = string, %d = integer, %f = float, %.2f = 2 decimal places
SELECT 
    name,
    salary,
    FORMAT_STRING('%s earns $%,d per year', name, salary) AS formatted_text,
    FORMAT_STRING('Salary: $%.2f', salary * 1.0) AS decimal_format
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- Practical: Generate CSV rows from data
SELECT 
    CONCAT_WS(',', 
        CAST(employee_id AS STRING), 
        name, 
        COALESCE(department, ''), 
        CAST(salary AS STRING)
    ) AS csv_row
FROM employees
ORDER BY employee_id;

-- Practical: Generate SQL INSERT statements dynamically
SELECT 
    FORMAT_STRING(
        'INSERT INTO employees VALUES (%d, ''%s'', ''%s'', %d);',
        employee_id, name, COALESCE(department, 'NULL'), salary
    ) AS insert_stmt
FROM employees
ORDER BY employee_id;
````

````sql
-- Map Manipulation: Advanced MAP functions
-- Build, combine, filter, and transform maps

-- MAP_FROM_ARRAYS: Create a map from two parallel arrays
SELECT 
    MAP_FROM_ARRAYS(
        ARRAY('name', 'dept', 'salary'),
        ARRAY('John', 'IT', 80000)
    ) AS emp_map;

-- MAP_FROM_ENTRIES: Create a map from an array of key-value pairs
SELECT 
    MAP_FROM_ENTRIES(ARRAY(
        STRUCT('a', 1),
        STRUCT('b', 2),
        STRUCT('c', 3)
    )) AS entry_map;

-- Create a temp view with maps for demos
CREATE OR REPLACE TEMP VIEW emp_attrs AS
SELECT 'John'  AS name, MAP('role', 'dev', 'level', '3', 'remote', 'true') AS attrs
UNION ALL
SELECT 'Mary',       MAP('role', 'analyst', 'level', '2', 'remote', 'false')
UNION ALL
SELECT 'David',      MAP('role', 'architect', 'level', '5', 'remote', 'true');

SELECT * FROM emp_attrs ORDER BY name;

-- MAP_CONCAT: Merge two or more maps (later values override earlier)
SELECT 
    name,
    MAP_CONCAT(attrs, MAP('status', 'active')) AS merged_attrs
FROM emp_attrs
ORDER BY name;

-- MAP_FILTER: Keep only key-value pairs matching a condition
SELECT 
    name,
    MAP_FILTER(attrs, (k, v) -> k = 'role' OR k = 'level') AS filtered_attrs
FROM emp_attrs
ORDER BY name;

-- TRANSFORM_VALUES: Apply a function to each value
SELECT 
    name,
    TRANSFORM_VALUES(attrs, (k, v) ->
        CASE WHEN k = 'level' THEN CAST(v AS INT) * 10000
             ELSE v END
    ) AS transformed_attrs
FROM emp_attrs
ORDER BY name;

-- TRANSFORM_KEYS: Apply a function to each key
SELECT 
    name,
    TRANSFORM_KEYS(attrs, (k, v) -> UPPER(k)) AS upper_keys
FROM emp_attrs
ORDER BY name;

-- MAP_VALUES / MAP_KEYS: Extract arrays from map
SELECT 
    name,
    MAP_KEYS(attrs)   AS all_keys,
    MAP_VALUES(attrs)  AS all_values
FROM emp_attrs
ORDER BY name;
````

````sql
-- Advanced JSON Queries: GET_JSON_OBJECT, JSON_ARRAY_LENGTH, JSON_OBJECT_KEYS, TO_JSON
-- Parse and navigate complex nested JSON structures

CREATE OR REPLACE TEMP VIEW json_events AS
SELECT * FROM VALUES
  ('{"user": "Alice", "action": "login", "meta": {"ip": "10.0.0.1", "device": "mobile"}}'),
  ('{"user": "Bob", "action": "purchase", "meta": {"ip": "10.0.0.2", "device": "desktop"}}'),
  ('{"user": "Carol", "action": "login", "meta": {"ip": "10.0.0.3", "device": "tablet"}}')
AS t(json_str);

-- GET_JSON_OBJECT: Extract a nested value by JSON path
SELECT 
    json_str,
    GET_JSON_OBJECT(json_str, '$.user')             AS user_name,
    GET_JSON_OBJECT(json_str, '$.action')           AS action,
    GET_JSON_OBJECT(json_str, '$.meta.ip')           AS ip_address,
    GET_JSON_OBJECT(json_str, '$.meta.device')       AS device_type
FROM json_events
ORDER BY user_name;

-- JSON_ARRAY_LENGTH: Count elements in a JSON array
SELECT 
    JSON_ARRAY_LENGTH('[1, 2, 3, 4, 5]') AS array_len,
    JSON_ARRAY_LENGTH('[]')               AS empty_len,
    JSON_ARRAY_LENGTH('[{"a": 1}]')        AS single_obj;

-- TO_JSON: Convert a struct/map to JSON string
SELECT 
    name,
    department,
    salary,
    TO_JSON(STRUCT(department, salary, name)) AS json_record
FROM employees
WHERE department IS NOT NULL
ORDER BY name;

-- FROM_JSON: Parse a JSON string into a typed struct
SELECT 
    FROM_JSON('{"name": "Alice", "age": 30}', 'STRUCT<name: STRING, age: INT>') AS parsed_struct;

-- Practical: Build a JSON summary per department
SELECT 
    department,
    TO_JSON(STRUCT(
        COUNT(*) AS emp_count,
        ROUND(AVG(salary)) AS avg_salary,
        COLLECT_LIST(name) AS employees
    )) AS dept_json
FROM employees
WHERE department IS NOT NULL
GROUP BY department
ORDER BY department;
````

````sql
-- Window Function Chaining: Combine multiple window functions in one query
-- Build complex analytics by layering ROW_NUMBER, LAG, SUM OVER, RANK, and more

-- Complete salary analysis: rank, running total, previous salary, and gap
SELECT 
    name,
    department,
    salary,
    -- Rank within department by salary
    RANK()       OVER (PARTITION BY department ORDER BY salary DESC) AS dept_rank,
    -- Global salary rank
    DENSE_RANK() OVER (ORDER BY salary DESC) AS global_rank,
    -- Running total across all rows ordered by salary
    SUM(salary)  OVER (ORDER BY salary DESC) AS running_total,
    -- Previous salary in the same department
    LAG(salary)  OVER (PARTITION BY department ORDER BY salary DESC) AS prev_dept_salary,
    -- Salary gap vs the person above (globally)
    salary - LAG(salary) OVER (ORDER BY salary DESC) AS salary_gap_above,
    -- Percent of department total
    ROUND(salary * 100.0 / SUM(salary) OVER (PARTITION BY department), 1) AS pct_of_dept_total,
    -- Cumulative average
    ROUND(AVG(salary) OVER (ORDER BY salary DESC ROWS BETWEEN UNBOUNDED PRECEDING AND CURRENT ROW)) AS cum_avg
FROM employees
WHERE department IS NOT NULL
ORDER BY salary DESC;

-- Month-over-month growth analysis (combining multiple windows)
CREATE OR REPLACE TEMP VIEW monthly_rev AS
SELECT * FROM VALUES
  ('2025-01', 100), ('2025-02', 120), ('2025-03', 115),
  ('2025-04', 140), ('2025-05', 135), ('2025-06', 160)
AS t(month, revenue);

SELECT 
    month,
    revenue,
    -- Previous month revenue
    LAG(revenue) OVER (ORDER BY month) AS prev_month,
    -- MoM change
    revenue - LAG(revenue) OVER (ORDER BY month) AS mom_change,
    -- MoM growth rate
    ROUND((revenue - LAG(revenue) OVER (ORDER BY month)) * 100.0 / 
          LAG(revenue) OVER (ORDER BY month), 1) AS mom_growth_pct,
    -- 3-month moving average
    ROUND(AVG(revenue) OVER (ORDER BY month ROWS BETWEEN 2 PRECEDING AND CURRENT ROW), 1) AS ma_3month,
    -- YTD cumulative
    SUM(revenue) OVER (ORDER BY month) AS ytd_total
FROM monthly_rev
ORDER BY month;
````
